```python
# Data Handling and Manipulation with Python

This project demonstrates core data handling and manipulation techniques using Python.
The focus is on loading data, cleaning missing values, filtering, grouping, and
extracting meaningful information from structured datasets.
```

```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv("Titanic.csv")
df.head()
```

```python
df.shape
df.columns
df.info()
df.describe()
```

```python
#Check missing values
df.isnull().sum()
```

```python
# Fill missing Age with median
df['Age']=df['Age'].fillna(df['Age'].median())
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
df.drop(columns=['Cabin'], inplace=True)
```

```python
df.isnull().sum()
```

```python
# Passengers older than 30
df[df['Age']>30]
```

```python
# Female passengers in 1st class
df[(df['Sex']=='female') & (df['Pclass']==1)]
```

```python
# Survival rate by gender
df.groupby('Sex')['Survived'].mean()
```

```python
# Survival rate by passenger class
df.groupby('Pclass')['Age'].mean()
```

```python
# Average age by class
df.groupby('Pclass')['Age'].mean()
```

```python
# Create New Features
# Family Size
df['FamilySize'] = df['SibSp'] + df['Parch'] +1
```

```python
# Fare perperson
df['FarePerPerson'] = df['Fare'] / df ['FamilySize']
```

```python
# Sort & Rank Data
```

```python
df.sort_values(by='Fare', ascending= False).head(10)
```

```python
## Key Insights

- Female passengers had a significantly higher survival rate than males.
- Passengers in higher classes showed better survival probabilities.
- Larger family sizes showed varying survival patterns.
- Fare per person provides better insight than total fare.
```
